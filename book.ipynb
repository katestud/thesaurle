```python
import wordfreq
common_words = wordfreq.top_n_list("en", 2_000)
common_words = set(common_words)
print(common_words)
```

```python
import requests
import os
from dotenv import load_dotenv

load_dotenv()

def get_synonyms_api_ninjas(word):
    url = "https://api.api-ninjas.com/v1/thesaurus?word={}".format(word)
    response = requests.get(url, headers={'X-Api-Key': os.getenv('API_NINJAS_KEY')})
    if response.status_code == requests.codes.ok:
        return response.json()["synonyms"]
    else:
        return None
```

```python
synonyms = set(get_synonyms_api_ninjas("mother"))
```

```python
x = synonyms.intersection(common_words)
print(synonyms)
print(x)
```
